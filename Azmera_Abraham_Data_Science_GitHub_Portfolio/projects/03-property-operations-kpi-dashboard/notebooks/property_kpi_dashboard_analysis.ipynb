# Property Operations KPI Dashboard Analysis

This notebook demonstrates portfolio-ready analysis using sample data.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/sample_property_kpis.csv')
df.head()
```

```python
df.describe(include='all')
```

```python

kpi_summary = df.groupby(['month', 'market']).agg(
    avg_occupancy=('occupancy_rate', 'mean'),
    avg_renewal=('renewal_rate', 'mean'),
    avg_delinquency=('delinquency_rate', 'mean'),
    total_revenue=('monthly_revenue', 'sum')
).reset_index()

kpi_summary.head()
```
