# OfferIntel Market Intelligence Analysis

This notebook demonstrates portfolio-ready analysis using sample data.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/sample_offer_data.csv')
df.head()
```

```python
df.describe(include='all')
```

```python

market_summary = df.groupby(['zip_code', 'price_band']).agg(
    transaction_count=('transaction_id', 'count'),
    avg_offer_to_list_ratio=('offer_to_list_ratio', 'mean'),
    avg_offer_count=('offer_count', 'mean'),
    avg_days_on_market=('days_on_market', 'mean'),
    inspection_waiver_rate=('inspection_waived', 'mean')
).reset_index()

market_summary.sort_values('avg_offer_to_list_ratio', ascending=False).head(10)
```
