# Housing Outcomes Dashboard Analysis

This notebook demonstrates portfolio-ready analysis using sample data.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/sample_housing_outcomes.csv')
df.head()
```

```python
df.describe(include='all')
```

```python

placement_by_barrier = df.groupby('primary_barrier').agg(
    participants=('participant_id', 'count'),
    placement_rate=('placed', 'mean'),
    retention_rate=('retained_90_days', 'mean')
).reset_index()

placement_by_barrier.sort_values('placement_rate')
```
